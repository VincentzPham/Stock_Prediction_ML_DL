```python
import os

# Danh sách các mã (stock codes)
stock_codes = ['AAPL', 'AMZN', 'AVGO', 'GOOG', 'META', 'MSFT', 'NVDA', 'SAP', 'TSLA', 'TSM']

# Danh sách các model
models = ['Multiple Linear Regression', 'Random Forrest', 'Decision Tree', 'ARIMA', 'SARIMA', 
          'Exponential Smoothing', 'Prophet', 'ANN', 'RNN', 'LSTM', 'BiLSTM', 'LSTM-GRU']

# Đường dẫn đến thư mục Stock
base_path = 'D:/Stock/'

# Các thư mục chính mà bạn đã tạo (Models, Prediction, Result, UI)
folders = ['Models', 'Prediction', 'Result', 'UI']

# Tạo subfolder cho mỗi mã trong mỗi thư mục chính, sau đó tạo các subfolder cho các model bên trong
for folder in folders:
    folder_path = os.path.join(base_path, folder)  # Đường dẫn đến folder chính
    if os.path.exists(folder_path):  # Kiểm tra folder chính có tồn tại không
        for stock in stock_codes:
            stock_folder_path = os.path.join(folder_path, stock)  # Đường dẫn đến thư mục mã
            if not os.path.exists(stock_folder_path):  # Nếu thư mục mã chưa tồn tại, tạo mới
                os.makedirs(stock_folder_path)
                print(f"Đã tạo thư mục: {stock_folder_path}")
            
            # Tạo subfolder cho các model bên trong thư mục mã
            for model in models:
                model_folder_path = os.path.join(stock_folder_path, model)  # Đường dẫn đến thư mục model
                if not os.path.exists(model_folder_path):  # Kiểm tra model có thư mục chưa
                    os.makedirs(model_folder_path)
                    print(f"Đã tạo thư mục model: {model_folder_path}")
                else:
                    print(f"Thư mục model {model_folder_path} đã tồn tại.")
    else:
        print(f"Thư mục {folder} không tồn tại.")
```
